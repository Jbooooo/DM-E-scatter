# Dark-matter–electron scattering in the Born approximation
This notebook demonstrates the helper functions provided by `dm_e_scatter` to compute
nonrelativistic scattering observables for a Yukawa potential.

```python
import numpy as np
import matplotlib.pyplot as plt

from dm_e_scatter import constants, kinematics, plots, units, yukawa
```

```python
# Benchmark parameters
m_chi = units.GeV_to_kg(0.05)  # 50 MeV
v_dm = units.km_per_s_to_m_per_s(220.0)
alpha_eff = 1e-2
mu = kinematics.reduced_mass(m_chi, constants.ELECTRON_MASS)

# Representative mediator masses
mediator_light = units.mass_kg_to_wave_number(units.GeV_to_kg(1e-7))
mediator_heavy = units.mass_kg_to_wave_number(units.GeV_to_kg(1.0))
mediator_scan = np.geomspace(mediator_light, mediator_heavy, 5)
```

## Angular differential cross section

```python
theta_grid = np.linspace(1e-4, np.pi, 400)
ax_angle = plots.plot_dsigma_dOmega(theta_grid, mu, v_dm, alpha_eff, mediator_light)
plots.save_figure(ax_angle, "../figures", "dsigma_dOmega")
ax_angle
```

## Recoil spectrum for light and heavy mediators

```python
t_max = kinematics.maximum_recoil_energy(mu, v_dm)
t_grid = np.linspace(0.0, t_max, 400)
ax_light = plots.plot_dsigma_dT(t_grid, mu, v_dm, alpha_eff, mediator_light)
ax_heavy = plots.plot_dsigma_dT(t_grid, mu, v_dm, alpha_eff, mediator_heavy)
plots.save_figure(ax_light, "../figures", "dsigma_dT_light")
plots.save_figure(ax_heavy, "../figures", "dsigma_dT_heavy")
ax_light, ax_heavy
```

## Mediator scan

```python
ax_scan = plots.scan_mediator_masses(mediator_scan, t_grid, mu, v_dm, alpha_eff)
plots.save_figure(ax_scan, "../figures", "dsigma_dT_scan")
ax_scan
```
